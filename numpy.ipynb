```python
import numpy as np

arr = np.array([1,3,5,7])
print('arr:', arr)
```

```python
arr1 = np.array([[1,3,5,7], [3,5,8,9]])
print('arr_matrix:', arr1)
arr1.shape
```

```python
# dtype
a =np.array([1,2,3])
print(a.dtype)
```

```python
# reshape
a = np.array([1,2,3,4,5,6])
b = a.reshape(2,3)
print('b_reshaped:', b)
```

```python
#astype
a = np.array([1,2,3])
b = a.astype(float)
print("float:", b)
```

```python
#np.zeros, np.ones => dummy data
z = np.zeros((1,3))
o = np.ones((2,3,4))
print("zeros:" ,z)
print("ones:", o)
```

```python
# ndim
a = np.array([[1,3], [4,5]])
print('ndim:', a.ndim)
```

```python
# arange
a = np.arange(1,6)
print('arrange:', a)
```

```python
# sum, mean, max, min
a = np.array([1,2,3,4])
print("sum:",a.sum())
print('mean:', a.mean())
print("max:",a.max())
print("min:", a.min())
```

```python
# flatten

a = np.array([[2,3], [3,4]])
print('flatten:', a.flatten())
```

```python
# T (Transport)
a = np.array([[1,2],[2,4]])
print('T:', a.T)
```

```python
# dot
a = np.array([[1,2],[2,4]])
b = np.array([[1,3],[5,4]])
print('dot:', a.dot(b))
```

```python
#random.rand => 0.0 va 1.0 / 0/0 included / 1.0 excluded
a = np.random.rand(2,4)
print('random_rand:', a )
```
